```python
# %% [markdown]
# # SolarMobil Heliark: ML Pipeline & Analysis Notebook
# Reproduces every attribute and metric on the Heliark race-engineering dashboard (KPIs, Powertrain, Battery, Solar, Track, Strategy, Digital Twin, System Health, Reports) from `heliark_telemetry_sim.csv`, then builds a proper ML pipeline on top.
#
# **Pipeline:** load → data quality → EDA (dashboard views) → physics digital twin → feature engineering → time-aware modelling (4 tasks) → model comparison → explainability → anomaly detection → save artefacts.
#
# Put this notebook next to `heliark_telemetry_sim.csv` (or in a folder with `data/heliark_telemetry_sim.csv`). Needs: pandas, numpy, matplotlib, seaborn, scikit-learn, scipy, joblib.

# %%
import warnings, joblib
from pathlib import Path
import numpy as np, pandas as pd, matplotlib.pyplot as plt, seaborn as sns
from sklearn.model_selection import TimeSeriesSplit, cross_val_score
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge
from sklearn.ensemble import RandomForestRegressor, HistGradientBoostingRegressor, IsolationForest
from sklearn.inspection import permutation_importance
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

warnings.filterwarnings("ignore")
plt.style.use("dark_background")
sns.set_palette(["#f5a623", "#33b5ff", "#46e07a", "#ff4d4d", "#c58cff"])
plt.rcParams.update({"figure.figsize": (12, 4), "axes.facecolor": "#0d131d", "figure.facecolor": "#070b12", "axes.grid": True, "grid.alpha": .25})
OR, BL, GR, RD = "#f5a623", "#33b5ff", "#46e07a", "#ff4d4d"
SEED = 42
OUT = Path("outputs"); (OUT / "models").mkdir(parents=True, exist_ok=True)

# %% [markdown]
# ## 1. Load data

# %%
PATH = next(p for p in ["heliark_telemetry_sim.csv", "data/heliark_telemetry_sim.csv"] if Path(p).exists())
df = pd.read_csv(PATH, parse_dates=["timestamp_ist"])
print(PATH, df.shape)
df.head()

# %% [markdown]
# ## 2. Data quality checks

# %%
dt = df.elapsed_s.diff().dropna()
print("Sample period (s): min %.2f  median %.2f  max %.2f" % (dt.min(), dt.median(), dt.max()))
print("Duplicate rows:", df.duplicated().sum(), "| Missing cells:", int(df.isna().sum().sum()))
const = [c for c in df.columns if df[c].nunique() <= 1]
print("Constant columns (ignored for ML):", const)
miss = df.isna().mean().sort_values(ascending=False); print(miss[miss > 0].head(10))
print("Fault codes:", df.fault_code.value_counts().to_dict(), "| XBee link OK %:", round(df.xbee_link_ok.mean() * 100, 2))
print("Max speed %.1f km/h (SEVC cap 60) -> compliant: %s" % (df.speed_kmph.max(), df.speed_kmph.max() <= 60))

# %%
GROUPS = {
 "KPI":        ["speed_kmph", "soc_pct", "motor_rpm", "pack_voltage_V", "motor_temp_C", "solar_array_power_W"],
 "Powertrain": ["motor_phase_current_A", "motor_torque_Nm", "controller_temp_C", "motor_efficiency_pct"],
 "Battery":    ["pack_voltage_V", "pack_current_A", "soc_pct", "pack_a_temp_C", "pack_b_temp_C", "soh_pct"],
 "Solar":      ["solar_array_power_W", "solar_array_voltage_V", "solar_array_current_A", "irradiance_ghi_Wm2", "solar_energy_cum_Wh"]}
pd.concat({g: df[c].describe().T[["mean", "std", "min", "max"]] for g, c in GROUPS.items()}).round(2)

# %% [markdown]
# ## 3. EDA mirroring the dashboard
# ### 3.1 Overview KPI cards (full-run traces)

# %%
fig, ax = plt.subplots(2, 3, figsize=(15, 7), sharex=True)
for a, c, col in zip(ax.ravel(), GROUPS["KPI"], [OR, GR, BL, GR, RD, OR]):
    a.plot(df.elapsed_s / 60, df[c], color=col, lw=.8); a.set_title(c)
for a in ax[1]: a.set_xlabel("minutes")
plt.tight_layout(); plt.show()

# %% [markdown]
# ### 3.2 Synchronized telemetry panel (speed, motor current, pack voltage, motor temp)

# %%
w = df[(df.elapsed_s >= 3600) & (df.elapsed_s < 3660)]   # a 60 s window, like the dashboard
fig, a1 = plt.subplots(figsize=(13, 4)); a2 = a1.twinx()
a1.plot(w.elapsed_s, w.speed_kmph, color=OR, label="Speed (km/h)"); a1.plot(w.elapsed_s, w.motor_phase_current_A, color=BL, label="Motor current (A)")
a2.plot(w.elapsed_s, w.pack_voltage_V, color=GR, label="Pack voltage (V)"); a2.plot(w.elapsed_s, w.motor_temp_C, color=RD, label="Motor temp (C)")
a1.set_xlabel("elapsed_s"); h1, l1 = a1.get_legend_handles_labels(); h2, l2 = a2.get_legend_handles_labels(); a1.legend(h1 + h2, l1 + l2, ncol=4, loc="upper center"); a2.grid(False); plt.show()

# %% [markdown]
# ### 3.3 Powertrain panel

# %%
fig, ax = plt.subplots(1, 3, figsize=(15, 4))
ax[0].scatter(df.motor_phase_current_A, df.motor_torque_Nm, s=3, c=df.motor_rpm, cmap="plasma"); ax[0].set(xlabel="Motor current (A)", ylabel="Torque (Nm)", title="Torque vs current (colour = RPM)")
ax[1].scatter(df.motor_rpm, df.motor_torque_Nm, s=3, color=OR); ax[1].set(xlabel="RPM", ylabel="Torque (Nm)", title="Torque-speed map")
ax[2].scatter(df.motor_mech_power_W, df.motor_efficiency_pct, s=3, color=GR); ax[2].set(xlabel="Mech power (W)", ylabel="Efficiency (%)", title="Efficiency map")
plt.tight_layout(); plt.show()
print("Mean efficiency %.1f %% | controller temp max %.1f C" % (df.motor_efficiency_pct.mean(), df.controller_temp_C.max()))

# %% [markdown]
# ### 3.4 Battery panel (pack voltage, current, SOC, temps, SOH)

# %%
fig, ax = plt.subplots(2, 2, figsize=(14, 7))
ax[0, 0].plot(df.elapsed_s / 60, df.soc_pct, color=GR); ax[0, 0].set_title("SOC (%)")
ax[0, 1].scatter(df.pack_current_A, df.pack_voltage_V, s=3, c=df.soc_pct, cmap="viridis"); ax[0, 1].set(title="Voltage sag: V vs I (colour = SOC)", xlabel="Current (A)", ylabel="V")
ax[1, 0].plot(df.elapsed_s / 60, df.pack_a_temp_C, color=BL, label="Pack A"); ax[1, 0].plot(df.elapsed_s / 60, df.pack_b_temp_C, color=RD, label="Pack B"); ax[1, 0].legend(); ax[1, 0].set_title("Pack temperatures (C)")
ax[1, 1].plot(df.elapsed_s / 60, df.cell_v_delta_mV, color=OR); ax[1, 1].set_title("Cell voltage spread (mV)")
plt.tight_layout(); plt.show()

# %% [markdown]
# ### 3.5 Solar panel

# %%
fig, ax = plt.subplots(1, 3, figsize=(15, 4))
sc = ax[0].scatter(df.irradiance_ghi_Wm2, df.solar_array_power_W, s=4, c=df.solar_panel_temp_C, cmap="inferno"); ax[0].set(xlabel="Irradiance (W/m2)", ylabel="Array power (W)", title="Power vs irradiance (colour = panel temp)"); plt.colorbar(sc, ax=ax[0])
ax[1].scatter(df.solar_array_voltage_V, df.solar_array_current_A, s=3, color=OR); ax[1].set(xlabel="V", ylabel="I", title="Array operating point")
ax[2].plot(df.elapsed_s / 60, df.solar_energy_cum_Wh / 1000, color=GR); ax[2].set(title="Cumulative solar energy (kWh)", xlabel="minutes")
plt.tight_layout(); plt.show()

# %% [markdown]
# ### 3.6 Track view: speed by lap distance and sector

# %%
LAP_LEN = df.lap_distance_m.max()
df["lap_frac"] = df.lap_distance_m / LAP_LEN
df["sector"] = pd.cut(df.lap_frac, [-.01, 1/3, 2/3, 1.01], labels=["S1", "S2", "S3"])
bins = np.linspace(0, 1, 101); df["bin"] = pd.cut(df.lap_frac, bins, labels=False, include_lowest=True)
heat = df.pivot_table(index="lap", columns="bin", values="speed_kmph", aggfunc="mean")
fig, ax = plt.subplots(1, 2, figsize=(15, 5), gridspec_kw={"width_ratios": [2.2, 1]})
im = ax[0].imshow(heat.values, aspect="auto", cmap="turbo", extent=[0, LAP_LEN, heat.index.max(), heat.index.min()]); plt.colorbar(im, ax=ax[0], label="km/h"); ax[0].set(xlabel="Lap distance (m)", ylabel="Lap", title="Speed heat-map (lap x distance)"); ax[0].grid(False)
df.groupby("sector").speed_kmph.mean().plot.bar(ax=ax[1], color=[OR, BL, RD]); ax[1].set(title="Average speed per sector (km/h)", ylabel="km/h")
plt.tight_layout(); plt.show()
print("Note: GPS columns are constant in this simulated file (lat %.4f), so the track is shown against lap distance." % df.gps_lat_deg.mean())

# %% [markdown]
# ### 3.7 Race strategy: lap table, energy per lap, SOC projection

# %%
laps = df.groupby("lap").agg(lap_time_s=("last_lap_time_s", "max"), soc_end=("soc_pct", "last"), energy_end=("net_battery_energy_Wh", "last"), avg_speed=("speed_kmph", "mean"), max_motor_temp=("motor_temp_C", "max"))
laps["energy_Wh"] = laps.energy_end.diff().fillna(laps.energy_end); laps["soc_used"] = (-laps.soc_end.diff()).fillna(100 - laps.soc_end)
laps = laps[laps.lap_time_s > 0]
fig, ax = plt.subplots(1, 3, figsize=(15, 4))
ax[0].bar(laps.index, laps.lap_time_s, color=OR); ax[0].set(title="Lap time (s)", xlabel="lap")
ax[1].bar(laps.index, laps.energy_Wh, color=BL); ax[1].set(title="Energy per lap (Wh)", xlabel="lap")
RACE_S = 7200; now = df.iloc[-1]; t_lap = laps.lap_time_s.median(); soc_lap = laps.soc_used.median()
ax[2].plot(df.elapsed_s / 60, df.soc_pct, color=OR, label="actual")
t0 = 30 * 60; r0 = df[df.elapsed_s <= t0].iloc[-1]; fin = r0.soc_pct - soc_lap * (RACE_S - t0) / t_lap
ax[2].plot([t0 / 60, RACE_S / 60], [r0.soc_pct, fin], "--", color=BL, label="projection from 30 min"); ax[2].legend(); ax[2].set(title="SOC and projected finish", xlabel="minutes")
plt.tight_layout(); plt.show()
prof = pd.DataFrame({"profile": ["Conservative", "Balanced", "Aggressive"], "lap_mult": [1.05, 1, .95], "energy_mult": [.93, 1, 1.1]})
prof["lap_time_s"] = t_lap * prof.lap_mult; prof["energy_Wh"] = laps.energy_Wh.median() * prof.energy_mult
prof["finish_soc_%"] = 100 - soc_lap * prof.energy_mult * RACE_S / prof.lap_time_s
display(prof.round(2)) if "display" in globals() else print(prof.round(2))

# %% [markdown]
# ### 3.8 System health

# %%
fig, ax = plt.subplots(1, 2, figsize=(14, 4))
for c, col in zip(["motor_temp_C", "controller_temp_C", "pack_a_temp_C", "mcu_temp_C"], [RD, OR, BL, GR]): ax[0].plot(df.elapsed_s / 60, df[c], color=col, label=c, lw=.9)
ax[0].axhline(90, ls=":", color="w"); ax[0].legend(); ax[0].set(title="Temperatures (C), dotted = 90 C motor warning", xlabel="minutes")
ax[1].plot(df.elapsed_s / 60, df.xbee_rssi_dBm, color=BL, lw=.7); ax[1].set(title="XBee RSSI (dBm)", xlabel="minutes")
plt.tight_layout(); plt.show()

# %% [markdown]
# ### 3.9 Correlation matrix of all dashboard attributes

# %%
DASH = list(dict.fromkeys(sum(GROUPS.values(), []))) + ["speed_target_kmph", "irradiance_ghi_Wm2", "pack_power_W"]
DASH = list(dict.fromkeys(DASH))
corr = df[DASH].corr()
plt.figure(figsize=(11, 9)); sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", center=0, annot_kws={"size": 7}, linewidths=.3); plt.title("Correlation of dashboard attributes"); plt.tight_layout(); plt.show()
pairs = corr.where(np.triu(np.ones(corr.shape), 1).astype(bool)).stack().sort_values(key=abs, ascending=False)
print("Strongest pairs:"); print(pairs.head(10).round(3))

# %% [markdown]
# ## 4. Physics digital twin (Digital Twin Validation panel)
# Twin from the engineering spec: `V = OCV(SOC) - I * R` for the pack, and commanded target speed for the speed channel. MAE / RMSE / R² as on the dashboard.

# %%
def rmse(a, b): return float(np.sqrt(mean_squared_error(a, b)))
s = df.soc_pct / 100
df["twin_v"] = 20 * (3.35 + .55 * s + .30 * s**3 - .20 * (1 - s)**4) - df.pack_current_A * df.pack_internal_resistance_mOhm / 1000
twin = {"Speed MAE (km/h)": mean_absolute_error(df.speed_kmph, df.speed_target_kmph), "Speed RMSE (km/h)": rmse(df.speed_kmph, df.speed_target_kmph),
        "Voltage MAE (V)": mean_absolute_error(df.pack_voltage_V, df.twin_v), "Voltage RMSE (V)": rmse(df.pack_voltage_V, df.twin_v), "Voltage R2": r2_score(df.pack_voltage_V, df.twin_v),
        "SEVC compliant": bool(df.speed_kmph.max() <= 60)}
print(pd.Series(twin).round(3).to_string())
fig, ax = plt.subplots(1, 3, figsize=(16, 4))
ax[0].plot(df.elapsed_s / 60, df.pack_voltage_V, color=GR, lw=.8, label="measured"); ax[0].plot(df.elapsed_s / 60, df.twin_v, color=BL, lw=.8, label="twin"); ax[0].legend(); ax[0].set(title="Pack voltage", xlabel="minutes")
ax[1].hist(df.twin_v - df.pack_voltage_V, bins=60, color=OR); ax[1].set(title="Twin voltage residual (V)")
ax[2].scatter(df.speed_target_kmph, df.speed_kmph, s=3, color=BL); ax[2].plot([0, 60], [0, 60], "w--"); ax[2].set(xlabel="target", ylabel="measured", title="Speed: target vs measured")
plt.tight_layout(); plt.show()

# %% [markdown]
# ## 5. Feature engineering
# Features are built from signals available *before* the target, using rolling windows (no future data). Splits are chronological, never shuffled.

# %%
d = df.sort_values("elapsed_s").reset_index(drop=True).copy()
d["accel_ms2"] = d.speed_kmph.diff().fillna(0) / 3.6 / d.elapsed_s.diff().fillna(1)
d["hour_frac"] = d.timestamp_ist.dt.hour + d.timestamp_ist.dt.minute / 60
for c in ["throttle_pct", "speed_kmph", "motor_phase_current_A"]:
    d[c + "_r10"] = d[c].rolling(10, min_periods=1).mean()
d["current_rms_30"] = np.sqrt((d.motor_phase_current_A**2).rolling(30, min_periods=1).mean())
d["twin_resid"] = d.pack_voltage_V - d.twin_v
d = d.ffill().bfill()
print(d.shape)

# %% [markdown]
# ## 6. Modelling utilities
# Chronological 80/20 split, `TimeSeriesSplit` cross-validation for model selection, three model families (Ridge, Random Forest, Gradient Boosting).

# %%
def models():
    return {"Ridge": Pipeline([("sc", StandardScaler()), ("m", Ridge(alpha=1.0))]),
            "RandomForest": RandomForestRegressor(n_estimators=150, max_depth=12, min_samples_leaf=5, n_jobs=-1, random_state=SEED),
            "GradBoost": HistGradientBoostingRegressor(max_depth=6, learning_rate=.06, max_iter=250, random_state=SEED)}

def run_task(name, X, y, base=None, test_frac=.2):
    """Fit all models; report CV MAE (train part) and hold-out MAE/RMSE/R2. `base` = naive/physics baseline on hold-out."""
    n = len(X); cut = int(n * (1 - test_frac)); Xtr, Xte, ytr, yte = X.iloc[:cut], X.iloc[cut:], y.iloc[:cut], y.iloc[cut:]
    rows, fitted, preds = [], {}, {}
    if base is not None:
        b = base.iloc[cut:]; rows.append(dict(task=name, model="Baseline", cv_MAE=np.nan, MAE=mean_absolute_error(yte, b), RMSE=rmse(yte, b), R2=r2_score(yte, b))); preds["Baseline"] = b
    for k, m in models().items():
        cv = -cross_val_score(m, Xtr, ytr, cv=TimeSeriesSplit(5), scoring="neg_mean_absolute_error").mean()
        m.fit(Xtr, ytr); p = pd.Series(m.predict(Xte), index=yte.index); fitted[k], preds[k] = m, p
        rows.append(dict(task=name, model=k, cv_MAE=cv, MAE=mean_absolute_error(yte, p), RMSE=rmse(yte, p), R2=r2_score(yte, p)))
    res = pd.DataFrame(rows).round(4); best = res[res.model != "Baseline"].sort_values("MAE").iloc[0].model
    return dict(name=name, res=res, fitted=fitted, preds=preds, best=best, Xte=Xte, yte=yte, Xtr=Xtr, ytr=ytr)

def report(r, unit=""):
    print(r["res"].to_string(index=False)); print("Best:", r["best"])
    yte, p = r["yte"], r["preds"][r["best"]]
    fig, ax = plt.subplots(1, 3, figsize=(16, 4))
    ax[0].plot(yte.index, yte, color="w", lw=.7, label="actual"); ax[0].plot(p.index, p, color=OR, lw=.7, label=r["best"]); ax[0].legend(); ax[0].set(title=r["name"] + ": hold-out", xlabel="sample")
    ax[1].scatter(yte, p, s=3, color=BL); lo, hi = yte.min(), yte.max(); ax[1].plot([lo, hi], [lo, hi], "w--"); ax[1].set(xlabel="actual", ylabel="predicted", title="Predicted vs actual")
    ax[2].hist(yte - p, bins=50, color=GR); ax[2].set(title="Residuals " + unit)
    plt.tight_layout(); plt.show()
    pi = permutation_importance(r["fitted"][r["best"]], r["Xte"], yte, n_repeats=5, random_state=SEED, scoring="neg_mean_absolute_error")
    imp = pd.Series(pi.importances_mean, index=r["Xte"].columns).sort_values().tail(10)
    imp.plot.barh(color=OR, figsize=(8, 3.5), title="Permutation importance (top 10)"); plt.tight_layout(); plt.show()
ALL = {}

# %% [markdown]
# ## 7. Task A: Battery power / energy model (feeds Pred. Energy per Lap)
# Predict pack power from driving inputs only (no electrical measurements, so no leakage). Baseline = spec resistive-force power estimate.

# %%
FA = ["speed_kmph", "speed_target_kmph", "accel_ms2", "throttle_pct", "throttle_pct_r10", "speed_kmph_r10", "slope_gradient_pct", "headwind_mps", "air_density_kgm3", "regen_active", "brake_pressure_front_MPa", "road_wheel_angle_deg", "imu_ay_g"]
base = d.resistive_force_total_N * d.speed_kmph / 3.6 / .85       # F*v / drivetrain eff
A = run_task("A: pack power (W)", d[FA], d.pack_power_W, base=base); ALL["A"] = A; report(A, "(W)")

# %% [markdown]
# ## 8. Task B: Thermal forecast, motor temperature 30 s ahead (System Health)
# Target is the temperature *change* over the next 30 s; the baseline is 'no change' (persistence).

# %%
H = 30
FB = ["motor_temp_C", "controller_temp_C", "motor_phase_current_A", "current_rms_30", "motor_torque_Nm", "motor_rpm", "speed_kmph", "ambient_temp_C", "pack_a_temp_C", "motor_phase_current_A_r10"]
yB = (d.motor_temp_C.shift(-H) - d.motor_temp_C); ok = yB.notna()
B = run_task("B: motor temp delta 30s (C)", d.loc[ok, FB], yB[ok], base=pd.Series(0.0, index=yB[ok].index)); ALL["B"] = B; report(B, "(C)")
best = B["preds"][B["best"]]; ytrue = d.motor_temp_C.shift(-H)[ok].loc[best.index]; now_t = d.motor_temp_C[ok].loc[best.index]
print("Motor temp 30 s ahead -> MAE %.3f C (persistence %.3f C)" % (mean_absolute_error(ytrue, now_t + best), mean_absolute_error(ytrue, now_t)))

# %% [markdown]
# ## 9. Task C: Solar array power (Solar System panel)
# Baseline = physics formula from the spec: `P = 700.6 * G/1000 * (1 - 0.004 (Tcell - 25)) * 0.98 * 0.97`.

# %%
FC = ["irradiance_ghi_Wm2", "solar_panel_temp_C", "ambient_temp_C", "cloud_cover_pct", "hour_frac", "humidity_pct", "wind_speed_mps"]
physC = 700.6 * d.irradiance_ghi_Wm2 / 1000 * (1 - .004 * (d.solar_panel_temp_C - 25)) * .98 * .97
C = run_task("C: solar power (W)", d[FC], d.solar_array_power_W, base=physC); ALL["C"] = C; report(C, "(W)")

# %% [markdown]
# ## 10. Task D: Hybrid digital twin (physics + ML residual)
# The physics twin leaves a systematic voltage error. A model learns that residual from state features; final prediction = twin + ML correction.

# %%
FD = ["soc_pct", "pack_current_A", "pack_a_temp_C", "pack_b_temp_C", "pack_c_rate", "soh_pct", "pack_internal_resistance_mOhm", "motor_phase_current_A_r10", "ambient_temp_C"]
D_ = run_task("D: twin residual (V)", d[FD], d.twin_resid, base=pd.Series(0.0, index=d.index)); ALL["D"] = D_; report(D_, "(V)")
cut = int(len(d) * .8); te = d.iloc[cut:]; corr_v = te.twin_v + D_["preds"][D_["best"]]
print("Voltage MAE on hold-out: physics twin %.3f V -> hybrid twin %.3f V" % (mean_absolute_error(te.pack_voltage_V, te.twin_v), mean_absolute_error(te.pack_voltage_V, corr_v)))
print("Voltage R2: physics %.4f -> hybrid %.4f" % (r2_score(te.pack_voltage_V, te.twin_v), r2_score(te.pack_voltage_V, corr_v)))

# %% [markdown]
# ## 11. Task E: Anomaly detection (Vehicle Health)
# Unsupervised Isolation Forest on health channels; compared with `fault_code` and temperature limits if faults exist.

# %%
FE = ["motor_temp_C", "controller_temp_C", "pack_a_temp_C", "pack_b_temp_C", "pack_voltage_V", "pack_current_A", "cell_v_delta_mV", "motor_efficiency_pct", "xbee_rssi_dBm", "mcu_temp_C"]
Z = StandardScaler().fit_transform(d[FE]); iso = IsolationForest(n_estimators=200, contamination=.01, random_state=SEED).fit(Z)
d["anom"] = iso.predict(Z) == -1; d["anom_score"] = -iso.score_samples(Z)
fig, ax = plt.subplots(2, 1, figsize=(14, 6), sharex=True)
ax[0].plot(d.elapsed_s / 60, d.motor_temp_C, color=RD, lw=.8); ax[0].scatter(d.elapsed_s[d.anom] / 60, d.motor_temp_C[d.anom], color="w", s=10, label="anomaly"); ax[0].legend(); ax[0].set_title("Motor temp with flagged anomalies")
ax[1].plot(d.elapsed_s / 60, d.anom_score, color=OR, lw=.7); ax[1].set(title="Anomaly score", xlabel="minutes"); plt.tight_layout(); plt.show()
print("Flagged %d samples (%.2f %%)" % (d.anom.sum(), d.anom.mean() * 100))
if d.fault_code.nunique() > 1: print(pd.crosstab(d.fault_code != 0, d.anom, rownames=["fault"], colnames=["anomaly"]))
else: print("fault_code is constant (no faults in this run), so the detector is unsupervised-only here.")
# what does a fault look like? standardised mean difference during faults vs normal
if (d.fault_code != 0).any():
    f = d.fault_code != 0; zd = ((d.loc[f, FE + ["xbee_latency_ms"]].mean() - d.loc[~f, FE + ["xbee_latency_ms"]].mean()) / d[FE + ["xbee_latency_ms"]].std()).abs().sort_values(ascending=False)
    print("\nChannels that differ most during faults (|z|):"); print(zd.head(4).round(2)); print("Fault vs XBee link down:"); print(pd.crosstab(d.fault_code, d.xbee_link_ok))

# %% [markdown]
# ## 12. Model comparison and saved artefacts

# %%
summary = pd.concat([v["res"] for v in ALL.values()], ignore_index=True)
summary.to_csv(OUT / "model_metrics.csv", index=False); display(summary) if "display" in globals() else print(summary)
plot_df = summary[summary.model != "Baseline"]
g = sns.catplot(data=plot_df, x="model", y="R2", col="task", kind="bar", height=3.4, aspect=.9, sharey=False); g.set_titles("{col_name}"); plt.show()
for k, v in ALL.items(): joblib.dump(v["fitted"][v["best"]], OUT / "models" / f"task_{k}_{v['best']}.joblib")
joblib.dump(iso, OUT / "models" / "task_E_isolation_forest.joblib")
laps.to_csv(OUT / "lap_summary.csv"); pd.Series(twin).to_csv(OUT / "twin_metrics.csv")
print("Saved to", OUT.resolve(), "->", sorted(p.name for p in (OUT / "models").iterdir()))

# %% [markdown]
# ## 13. How this maps to the dashboard
# | Dashboard panel | Notebook section |
# |---|---|
# | Overview KPI cards, Synchronized Telemetry | 3.1, 3.2 |
# | Powertrain / Battery / Solar side panels | 3.3, 3.4, 3.5 |
# | Track map, sector colours | 3.6 |
# | Race Strategy (target lap, energy/lap, finish SOC) | 3.7, Task A |
# | Digital Twin Validation (MAE, RMSE, R²) | 4, Task D |
# | System Health | 3.8, Task B, Task E |
# | Reports (lap summary, CSV export) | 3.7, 12 |
#
# **Notes:** results come from a simulated run, so model scores describe how well the pipeline recovers the simulator's relationships, not real-track accuracy. Re-run on real telemetry before trusting any number.
```
